```python
price = d['Price (in rupees)']
```

```python
p=price.dropna()
p
```

```python
plt.boxplot(p)
```

```python
Q1 = p.quantile(0.25)
Q2 = p.quantile(0.5)
Q3 = p.quantile(0.75)
Q1, Q2, Q3
```

```python
IQR=Q3-Q1
IQR
```

```python
LL = Q1-1.5 * IQR
UL = Q3+1.5 * IQR
print(LL,",",UL)
```

```python
outliers = p[(p<LL) | (p>UL)]
outliers
```
